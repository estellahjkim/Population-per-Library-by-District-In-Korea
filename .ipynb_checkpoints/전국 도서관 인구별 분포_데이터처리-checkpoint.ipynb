```python
import pandas as pd
# pd.set_option('display.max_rows', None)
```

# 도서관 데이터 전처리

```python
small_library = pd.read_csv('전국작은도서관통계_20231231.csv', encoding = 'cp949')
small_library
```

```python
# 아파트 작은 도서관 출입 불가인 곳도 있음
# 방문자수 0인 곳 거르기

daegu_small_library = small_library[small_library['행정구역'] == '대구']

daegu_small_library[daegu_small_library['시군구'] == '동구']
```

```python
public_library = pd.read_csv('전국공공도서관정보_20231231.csv', encoding = 'cp949')
public_library
```

```python
# public_library와 small_library 합치기

public_library = public_library.drop(columns = ['평가년도','도서관구분', '도서관코드', '대출자수', '대출권수', '도서예산(자료구입비)'])
public_library = public_library.rename(columns={'장서수(인쇄)': '장서수'})

small_library = small_library.drop(columns = ['도서관구분', '방문자수', '도서예산자료 구입비(도서_비도서_전자자료_연속간행물) (원)'])

library = pd.concat([public_library, small_library])
library = library.drop(columns = ['사서수', '평가년도'])
library
```

# 시군구별 인구 데이터 전처리

```python
# 시군구 별 인구수 받아오기

population = pd.read_csv('202506_202506_연령별인구현황_월간.csv', encoding = 'cp949', )
population = population.iloc[:, [0, 1]]

population[['행정구역', '시군구']] = population['행정구역'].str.split(' ', n=1, expand=True)
```

```python
population['시군구'] = population['시군구'].str.replace(r'\s*\(.*\)', '', regex=True)

# 세종특별자치시 제외

sejong = population[population['행정구역']=='세종특별자치시']

sejong = sejong.iloc[[0]]

population = population[population['시군구']!='']
population = pd.concat([population, sejong], ignore_index=True)
population = population[['행정구역', '시군구', '2025년06월_계_총인구수']]

population = population.replace(to_replace=['서울특별시','부산광역시','대구광역시','인천광역시','광주광역시','대전광역시','울산광역시','경기도','강원특별자치도','충청북도','충청남도','전북특별자치도','전라남도','경상북도','경상남도','제주특별자치도','세종특별자치시'], value=['서울','부산','대구','인천','광주','대전','울산','경기','강원','충북','충남','전북','전남','경북','경남','제주','세종'])
population.loc[population['행정구역']=='세종', '시군구'] = '세종특별자치시'
population
```

# group by

```python
count = library.groupby(['행정구역', '시군구']).count()
sum = library.groupby(['행정구역', '시군구']).sum()
```

```python
count = count[count.columns[[0]]]
sum = sum[sum.columns[[1]]]
```

```python
library_population = count.join(sum)
library_population = library_population.rename(columns={'도서관명': '도서관수'})
library_population
```

```python
library_population_per_population = pd.merge(left=library_population, right=population, how='left', on=['행정구역', '시군구'])
library_population_per_population
```

# 분석

```python
library_population_per_population.info()
```

```python
# 총인구수 object에서 int으로 변환

lst = []

for i in range(library_population_per_population.shape[0]):
    
    value = library_population_per_population['2025년06월_계_총인구수'][i]
    if isinstance(value, str):
        value_change = int(value.replace(',',''))
    elif pd.isna(value):
        value_change = None  # 혹은 0, 또는 원하는 값
    else:
        value_change = int(value)
        
    lst.append(value_change)

library_population_per_population['2025년06월_계_총인구수'] = lst
library_population_per_population.dropna()
library_population_per_population
```

```python
# 시군구 별 도서관 당 인구 수

library_population_per_population['도서관 개당 인구수'] = library_population_per_population['2025년06월_계_총인구수'] / library_population_per_population['도서관수']
library_population_per_population
```

```python
# library_population_per_population를 csv로 변환

library_population_per_population.to_csv('library_population_per_population.csv', index=False, encoding='cp949')
```
